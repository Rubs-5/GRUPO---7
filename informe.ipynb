#### est: Adolfo de la torre contreras 
#### Clases: transaccion y cuenta 


## 1. Descripción de las clases

### 1.1 Clase `Transaccion` — `cuenta.py`
Representa un movimiento de dinero (depósito, retiro, transferencia, interés o comisión).

| Atributo | Descripción |
|---|---|
| `__tipo` | Tipo de movimiento ("Deposito", "Retiro", "Transferencia enviada"…) |
| `__monto` | Monto del movimiento |
| `__saldo_resultante` | Saldo que quedó después del movimiento |

**Métodos:** `get_tipo()`, `get_monto()`, `get_saldo_resultante()`, `__str__()`.

### 1.2 Clase `Cuenta` (clase padre) — `cuenta.py`

| Atributo | Descripción |
|---|---|
| `contador` (de clase) | Contador compartido para generar números de cuenta únicos |
| `__numero` | Número de cuenta (por ejemplo "001-00001") |
| `__titular` | Objeto `Cliente` dueño de la cuenta |
| `__saldo` | Saldo actual (privado y **sin setter**) |
| `__movimientos` | Lista de objetos `Transaccion` |

| Método | Qué hace |
|---|---|
| `__init__(titular, saldo_inicial=0)` | Constructor. Si hay saldo inicial, lo registra como "Apertura de cuenta" |
| `get_numero()`, `get_titular()`, `get_saldo()`, `get_movimientos()`, `get_tipo()` | Getters |
| `__registrar(tipo, monto)` | **Privado**: guarda una `Transaccion` en el historial |
| `depositar(monto, tipo="Deposito")` | Suma dinero si el monto es mayor que 0 |
| `puede_retirar(monto)` | Regla general: no retirar más de lo que hay |
| `retirar(monto, tipo="Retiro")` | Resta dinero si `puede_retirar()` lo permite |
| `cierre_de_mes()` | Vacío en el padre; cada subclase lo sobrescribe |
| `estado_de_cuenta()` | Imprime todos los movimientos y el saldo |
| `__str__()` | Texto resumido de la cuenta |
