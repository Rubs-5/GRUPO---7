## 4. Módulo de Interfaz de Usuario y Menú Principal (`main.py` / Consola)

### 4.1. Descripción General
Este componente actúa como la capa de presentación e interacción con el usuario. Proporciona un menú interactivo en consola estructurado en bucle, permitiendo gestionar las entidades bancarias (clientes, cuentas corrientes y cuentas de ahorros) y ejecutar transacciones financieras (depósitos, retiros y transferencias) de forma controlada.

---

### 4.2. Funciones de Validación y Utilidades de Entrada
Para evitar excepciones en tiempo de ejecución por datos mal formateados, se diseñaron funciones defensivas de lectura:

| Función | Descripción y Validación |
| :--- | :--- |
| `leer_monto(mensaje)` | Valida que la entrada del usuario corresponda a un número real positivo. Emplea `.replace(".", "", 1).isdigit()` para admitir decimales válidos y retorna `-1` ante entradas no numéricas. |
| `leer_opcion(mensaje)` | Valida que la selección ingresada sea un número entero mediante `.isdigit()`, devolviendo `-1` en caso de error. |
| `mostrar_titulo(banco)` | Formatea e imprime el encabezado del banco utilizando el método `banco.get_nombre().upper()`. |
| `cargar_datos_de_ejemplo(banco)` | Facilita pruebas y demostraciones poblando el sistema con clientes, cuentas de ahorros y cuentas corrientes con saldos iniciales. |

---

### 4.3. Funciones de Operaciones e Integración con POO
Cada opción del menú delega la lógica de negocio a los objetos instanciados, protegiendo la encapsulación de los datos:

* **Registro y Apertura:**
  * `registrar_cliente(banco)`: Captura DNI, nombres, apellidos y teléfono para invocar el registro en la instancia del banco.
  * `abrir_cuenta(banco)`: Solicita el DNI del titular, selecciona el subtipo de cuenta (1: Ahorros, 2: Corriente) y asigna el depósito inicial validado.
* **Transacciones Financieras:**
  * `depositar(banco)` y `retirar(banco)`: Buscan la cuenta por su identificador. Si existe, invocan los métodos polimórficos de la cuenta (`depositar()` o `retirar()`), informando el saldo actualizado mediante `cuenta.get_saldo()`.
  * `transferir(banco)`: Solicita cuenta de origen, cuenta de destino y monto, delegando la transacción directamente al método `banco.transferir()`.
* **Consultas y Cierres:**
  * `estado_de_cuenta(banco)`: Permite imprimir el historial detallado de movimientos de una cuenta específica a través de su método `estado_de_cuenta()`.
  * Operaciones bancarias globales: Opciones directas para `banco.mostrar_clientes()`, `banco.cierre_de_mes()` (cálculo de intereses y comisiones) y `banco.reporte_general()`.

---

### 4.4. Flujo de Control Principal (`main`)
La función `main()` implementa un bucle `while opcion != 0` que mantiene el sistema en ejecución continua hasta que el usuario decida salir (opción `0`). El menú evalúa la entrada a través de condicionales estructuradas (`if-elif-else`), manejando opciones fuera de rango con mensajes informativos sin interrumpir la ejecución del programa.
