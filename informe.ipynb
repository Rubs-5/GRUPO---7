## 1. Introducción y Escenario Simulado  
El presente trabajo detalla el diseño, la lógica y la implementación de un Sistema Bancario. Este proyecto consiste en una simulación desarrollada en Python que modela el funcionamiento interno y las reglas de negocio de una institución financiera real. El objetivo principal de este desarrollo es construir un programa robusto, seguro y escalable, aplicando los conceptos fundamentales de la Programación Orientada a Objetos (POO).

Dentro de este escenario simulado, el sistema actúa como un administrador central que gestiona la interacción entre las personas y sus productos financieros. Para ello, el programa permite registrar usuarios (Clientes) y vincularlos con diferentes productos bancarios (Cuentas). Con el fin de reflejar la complejidad del mundo real, se han diseñado dos tipos de productos con reglas de negocio distintas:

- **Cuentas de Ahorros:** Orientadas a la rentabilidad, generando intereses a favor del cliente al finalizar el mes.

- **Cuentas Corrientes:** Orientadas a la liquidez, permitiendo al cliente realizar retiros mayores a su saldo (sobregiro), a cambio del cobro de una comisión de mantenimiento mensual.

Finalmente, el sistema cuenta con un menú interactivo por consola que permite procesar y registrar el historial de operaciones diarias, tales como depósitos, retiros, transferencias seguras entre distintas cuentas y la visualización detallada del estado de cuenta de cada usuario.

**Análisis de Relaciones entre Clases:**
El diseño del sistema se sostiene sobre relaciones fundamentales de la Programación Orientada a Objetos que permiten el flujo de información:

1. **Herencia:**
La arquitectura centraliza la lógica común de las cuentas bancarias en una clase padre llamada `Cuenta`. De esta clase derivan dos clases hijas: `CuentaAhorros` y `CuentaCorriente`. Ambas heredan los atributos básicos (titular, número, saldo) y métodos (depositar, obtener saldo), pero implementan su propio comportamiento especializado para retiros y comisiones.

2. **Asociación y Composición:**

- Un objeto `Banco` tiene listas maestras que almacenan múltiples objetos de tipo `Cliente` y `Cuenta`.

- Un objeto `Cliente` tiene una lista personal que almacena objetos de tipo `Cuenta` que le pertenecen.

- Un objeto `Cuenta` tiene una lista de objetos `Transaccion` que actúan como el historial de movimientos financieros (estado de cuenta).

Esta separación de responsabilidades asegura que cada clase tenga un propósito único.

## 3. Conceptos de POO Aplicados
El desarrollo de este sistema bancario no es un simple conjunto de funciones aisladas, sino una arquitectura sólida basada en los pilares fundamentales de la Programación Orientada a Objetos. A continuación, se detalla cómo se aplicó cada concepto en el código:

### 3.1. Encapsulamiento (Protección de datos)
Se aplicó un encapsulamiento estricto en todas las clases para proteger la integridad del sistema. Todos los atributos de las clases fueron declarados como privados (utilizando el prefijo `__`, por ejemplo: `__saldo`, `__dni`, `__movimientos`).

- **Acceso controlado:** Para leer la información, se implementaron métodos Getters específicos (como `get_saldo()`).

- **Modificación segura:** La modificación de atributos clave está restringida. Por ejemplo, el atributo `__saldo` **no tiene un método setter**; el dinero solo puede modificarse pasando por las reglas de validación de los métodos `depositar()` y `retirar()`, impidiendo que alguien altere el saldo directamente. Otro ejemplo es `set_telefono()` en la clase `Cliente`, el cual valida que el ingreso sea estrictamente de 9 dígitos numéricos antes de guardar el cambio.

### 3.2. Herencia (Reutilización de código)
Para evitar la duplicación de código y establecer una jerarquía lógica, se creó una clase padre llamada `Cuenta`.

Esta clase base contiene toda la lógica general de una cuenta bancaria (generación de número de cuenta, registro de historial, métodos de depósito).

A partir de ella, se crearon dos clases hijas: `CuentaAhorros` y `CuentaCorriente`. Estas subclases heredan automáticamente todo el comportamiento de la clase padre, pero añaden sus propios atributos (como tasas de interés o líneas de sobregiro) y reglas específicas.

### 3.3. Polimorfismo (Comportamiento dinámico)
El polimorfismo es una de las herramientas utilizadas en la clase `Banco`. Existen dos ejemplos claros en nuestro sistema:

- **El método `cierre_de_mes()`:** En la clase `Banco`, el sistema recorre la lista maestra de todas las cuentas y ejecuta `cuenta.cierre_de_mes()`. El banco no necesita preguntar de qué tipo es la cuenta; gracias al polimorfismo, la `CuentaAhorros` sabe que debe sumarse intereses, y la `CuentaCorriente` sabe que debe restarse una comisión. Cada objeto reacciona de forma diferente al mismo llamado.

- **El método `puede_retirar(monto)`:** La clase padre dicta la regla general de que no se puede retirar más de lo que hay en el saldo, pero la `CuentaCorriente` sobrescribe (hace un override) de este método para permitir retiros mayores al saldo utilizando su límite de sobregiro.

### 3.4. Abstracción
Se ocultó la complejidad de las operaciones financieras internas. Cuando el usuario (o la clase `Banco`) invoca el método `transferir()`, no necesita preocuparse por cómo se registran las transacciones individuales internamente en el historial; simplemente llama al método, y el sistema se encarga de abstraer la complejidad coordinando los retiros, depósitos y registros automáticos de cada objeto `Transaccion`.

## 4. Diccionario y Descripción de Clases


### 4.1 Clase Cliente

La clase `Cliente` tiene como propósito representar computacionalmente a una persona que utiliza los servicios del banco. Esta clase hace un uso del concepto de **encapsulamiento** para proteger la integridad de los datos personales y establece una relación de asociación, ya que un **"Cliente tiene Cuentas"**.

**Atributos:**  
Todos los atributos de esta clase están **encapsulados** (llevan el prefijo `__`) para evitar que sean modificados directamente desde fuera de la clase:
*   `__dni`: Almacena el Documento Nacional de Identidad del usuario.
*   `__nombre`: Almacena únicamente los nombres del cliente.
*   `__apellido`: Almacena los apellidos del cliente.
*   `__telefono`: Almacena el número de contacto. Por defecto, si no se proporciona, se inicializa como una cadena de texto vacía `""`.
*   `__cuentas`: Es una lista inicialmente vacía que sirve para almacenar los objetos de tipo `Cuenta` (como cuentas de ahorro o corrientes) que le pertenezcan a este usuario.

**Constructor:**
*   `__init__(self, dni, nombre, apellido, telefono="")`: Es el método inicializador que recibe los datos básicos del cliente al momento de crearlo en el sistema y los asigna a sus respectivos atributos privados. También inicializa la lista vacía de cuentas.

**Métodos de Acceso (Getters):**  
Dado que los atributos son privados, estos métodos permiten leer su valor de forma segura:
*   `get_dni(self)`: Retorna el DNI del cliente.
*   `get_nombre_completo(self)`: Retorna una cadena de texto que une el atributo `__nombre`, un espacio en blanco y el atributo `__apellido` para facilitar la visualización del nombre completo.
*   `get_telefono(self)`: Retorna el número de teléfono actual del cliente.
*   `get_cuentas(self)`: Retorna la lista completa de objetos `Cuenta` que posee el cliente.

**Métodos de Modificación (Setters):** 
*   `set_telefono(self, nuevo_telefono)`: Es el único setter de la clase. Antes de modificar el atributo `__telefono`, realiza una validación estricta: verifica que el valor ingresado contenga únicamente números (`isdigit()`) y que su longitud sea exactamente de 9 caracteres. Si no cumple la condición, rechaza el cambio e imprime un mensaje de error.

**Métodos de Operación y Visualización:**
*   `agregar_cuenta(self, cuenta)`: Recibe un objeto de tipo `Cuenta` y lo añade (`append`) a la lista privada `__cuentas` del cliente.
*   `saldo_total(self)`: Calcula el patrimonio del cliente. Utiliza un bucle `for` para recorrer todas las cuentas en `__cuentas`, obtiene el saldo individual de cada una usando `get_saldo()` y las suma en una variable total que retorna al final.
*   `mostrar(self)`: Es un método de visualización detallada. Imprime en consola el nombre completo, DNI y teléfono del cliente. Luego, si la lista de cuentas está vacía, indica que no tiene cuentas; de lo contrario, recorre la lista imprimiendo la información de cada cuenta y finaliza mostrando el saldo total formateado a dos decimales.
*   `__str__(self)`: Es un método especial de Python que devuelve una representación resumida del cliente en forma de texto, mostrando su DNI, su nombre completo y la cantidad exacta de cuentas que posee (calculada con `len(self.__cuentas)`).

### 4.2 Clase Banco

La clase `Banco` actúa como el motor o administrador central de todo el sistema bancario. Su función principal es gestionar todas las operaciones que ocurren entre las distintas personas e instancias del sistema, aplicando conceptos avanzados de POO como **relaciones entre clases** (listas de objetos) y **polimorfismo**.

**Atributos:**  
Para mantener el control absoluto sobre los registros financieros, el banco encapsula sus datos más importantes:
*   `__nombre`: Almacena el nombre comercial de la institución financiera.
*   `__clientes`: Es una lista que funciona como la base de datos central de usuarios, almacenando únicamente objetos de tipo `Cliente`.
*   `__cuentas`: Es una lista maestra que registra todas las cuentas creadas en la institución (objetos de tipo `CuentaAhorros` y `CuentaCorriente`), independientemente de a qué cliente pertenezcan.

**Constructor:**
*   `__init__(self, nombre)`: Al instanciar el banco, recibe su nombre comercial e inicializa las listas `__clientes` y `__cuentas` totalmente vacías, listas para empezar a registrar información.

**Métodos de Acceso (Getters):**
*   `get_nombre(self)`: Retorna el nombre del banco.
*   `get_clientes(self)`: Retorna la lista maestra de todos los clientes registrados.

**Métodos de Gestión de Clientes:**
*   `buscar_cliente(self, dni)`: Recorre la lista `__clientes` buscando una coincidencia exacta con el DNI proporcionado. Si lo encuentra, retorna el objeto `Cliente`; si no, devuelve `None`.
*   `registrar_cliente(self, dni, nombre, apellido, telefono="")`: Valida estrictamente que el DNI ingresado tenga 8 dígitos numéricos. Luego, utiliza `buscar_cliente()` para asegurarse de que no exista ya un usuario con ese DNI. Si pasa las validaciones, instancia un nuevo objeto `Cliente`, lo agrega a la lista `__clientes` y lo retorna.

**Métodos de Gestión de Cuentas:**
*   `buscar_cuenta(self, numero)`: Funciona igual que la búsqueda de clientes, pero recorre la lista `__cuentas` buscando coincidencia con el número de cuenta.
*   `abrir_cuenta(self, dni, tipo, saldo_inicial=0)`: Es un método fundamental. Primero verifica que el cliente exista. Dependiendo del argumento `tipo`, instancia una `CuentaAhorros` (si el tipo es 1) o una `CuentaCorriente` (si el tipo es 2). Una vez creado el objeto, lo vincula agregándolo tanto a la lista personal del cliente (`cliente.agregar_cuenta()`) como a la lista general del banco (`self.__cuentas`).

**Métodos de Operaciones Financieras:**
*   `transferir(self, numero_origen, numero_destino, monto)`: Maneja la lógica de enviar dinero entre dos cuentas. Verifica que ambas cuentas existan y no sean la misma. La regla de oro aquí es: intenta ejecutar el retiro en la cuenta de origen, y **solo si esta operación fue exitosa** (es decir, había saldo suficiente o sobregiro permitido), procede a ejecutar el depósito en la cuenta de destino.
*   `cierre_de_mes(self)`: **Polimorfismo**. El banco recorre su lista maestra de cuentas y llama al método `cuenta.cierre_de_mes()` para cada una. El banco no necesita saber qué tipo de cuenta es; gracias al polimorfismo, cada objeto "sabe" cómo comportarse (la cuenta de ahorros se sumará intereses, mientras que la corriente se restará la comisión de mantenimiento).

**Métodos de Reportes:**
*   `mostrar_clientes(self)`: Recorre la lista `__clientes` ejecutando el método `mostrar()` de cada uno, generando un reporte impreso de todos los usuarios y sus cuentas asociadas.
*   `reporte_general(self)`: Itera sobre todas las cuentas existentes, suma el dinero de cada una y genera un resumen estadístico mostrando la cantidad total de clientes, cuentas abiertas y el patrimonio total gestionado por el banco.

### 4.3 Clase Transaccion
Esta clase tiene la responsabilidad de representar y registrar un único movimiento de dinero ocurrido en el sistema (como un depósito, retiro, transferencia, interés o comisión). Actúa como el historial contable del banco.

**Atributos (Privados):**

- `__tipo`: Almacena el concepto o tipo de movimiento (ej. "Depósito", "Retiro", "Transferencia enviada").

- `__monto`: Guarda la cantidad de dinero involucrada en la operación.

- `__saldo_resultante`: Registra una "fotografía" del saldo que quedó en la cuenta exactamente después de realizarse este movimiento.

**Métodos principales:**

- **Getters:** `get_tipo()`, `get_monto()`, `get_saldo_resultante()`. Permiten leer la información de la transacción sin alterarla.

- `__str__(self)`: Retorna una cadena de texto formateada para que, al imprimir la transacción, se vea tabulada como un registro de estado de cuenta.

### 4.4 Clase Cuenta (Clase Padre)
Es la superclase o clase base que define lo que tienen en común todas las cuentas bancarias del sistema. No se instancia directamente para el uso final, sino que sirve como plantilla para las cuentas de ahorro y corrientes.

**Atributos:**

- `contador` (Atributo de clase): Es un contador compartido por todos los objetos que sirve para generar números de cuenta correlativos y únicos de forma automática.

- `__numero`: Almacena el código único de la cuenta (por ejemplo, "001-00001").

- `__titular`: Almacena el objeto `Cliente` al que le pertenece la cuenta.

- `__saldo`: Almacena el dinero actual. Es estrictamente privado y no tiene setter.

- `__movimientos`: Lista que guarda objetos de tipo `Transaccion`.

**Métodos principales:**

- `__init__(self, titular, saldo_inicial=0)`: Constructor que inicializa la cuenta. Si se detecta un `saldo_inicial` mayor a 0, automáticamente registra una transacción de "Apertura de cuenta".

- `__registrar(self, tipo, monto)`: Método Privado. Su función interna es crear un nuevo objeto `Transaccion` y añadirlo a la lista de `__movimientos`.

- `depositar(self, monto)`: Valida que el monto sea positivo, lo suma al `__saldo` y llama al método privado `__registrar()`.

- `puede_retirar(self, monto)`: Contiene la regla de validación general (el monto no puede superar el saldo actual).

- `retirar(self, monto)`: Verifica si `puede_retirar()` es True. Si lo es, resta el monto del saldo y registra la transacción.

- `estado_de_cuenta(self)`: Método de visualización que imprime los datos del titular y recorre con un bucle `for` la lista de `__movimientos`, mostrándolos en formato de boleta.

- `cierre_de_mes(self)`: Método diseñado para ser sobrescrito (Override) obligatoriamente por las clases hijas.

---

### 4.5 Clase CuentaAhorros (Subclase)
Esta clase hereda de `Cuenta` y representa un producto financiero diseñado para rentabilizar el dinero del usuario. Su lógica de negocio premia el ahorro generando intereses, pero restringe la liquidez limitando la cantidad de retiros que el usuario puede hacer.

**Atributos específicos:**

- `LIMITE_RETIROS` (Atributo de clase): Define el máximo de retiros permitidos por mes (configurado a 3).

- `__tasa_interes` (Privado): Porcentaje de ganancia mensual (por defecto 0.01 o 1%).

- `__retiros_del_mes` (Privado): Contador interno que lleva el registro de cuántas veces ha sacado dinero el usuario en el periodo actual.

**Métodos sobrescritos (Polimorfismo / Override):**

- `puede_retirar(self, monto)`: Antes de verificar el saldo, evalúa si el contador `__retiros_del_mes` ha alcanzado el límite. Si lo alcanzó, bloquea la operación. Si no, invoca la validación de la clase padre (`uenta.puede_retirar()`).

- `retirar(self, monto, tipo)`: Intenta realizar el retiro usando la lógica de la clase padre. Si es exitoso, incrementa el contador `__retiros_del_mes` en 1.

- `cierre_de_mes(self)`: Calcula el interés ganado multiplicando el saldo actual por la tasa de interés. Si la ganancia es mayor a 0, hace un depósito automático ("Intereses del mes") y finalmente reinicia el contador de retiros a 0 para el nuevo mes.

---

### 4.6 Clase CuentaCorriente (Subclase)
Esta clase también hereda de `Cuenta`, pero está orientada a la liquidez total. Permite al cliente gastar dinero que no tiene a través de una línea de crédito, a cambio del cobro de una comisión mensual.

**Atributos específicos:**

- `COMISION_MANTENIMIENTO` (Atributo de clase): Fija el cobro mensual del banco (configurado a S/ 10).

- `__limite_sobregiro` (Privado): Establece cuánto dinero en negativo o "prestado" puede sacar el titular (por defecto S/ 500).

**Métodos sobrescritos (Polimorfismo / Override):**

- `puede_retirar(self, monto)`: Modifica radicalmente la regla general. Permite aprobar retiros incluso si el monto es mayor al saldo actual, siempre y cuando la diferencia no supere el margen de `__limite_sobregiro`.

- `cierre_de_mes(self)`: Realiza un cargo automático llamando al método `retirar()` por el monto de la comisión de mantenimiento.

### 5.1. Descripción General
Este componente actúa como la capa de presentación e interacción con el usuario. Proporciona un menú interactivo en consola estructurado en bucle, permitiendo gestionar las entidades bancarias (clientes, cuentas corrientes y cuentas de ahorros) y ejecutar transacciones financieras (depósitos, retiros y transferencias) de forma controlada.

### 5.2. Funciones de Validación y Utilidades de Entrada
Para evitar excepciones en tiempo de ejecución por datos mal formateados, se diseñaron funciones defensivas de lectura:

| Función | Descripción y Validación |
| :--- | :--- |
| `leer_monto(mensaje)` | Valida que la entrada del usuario corresponda a un número real positivo. Emplea `.replace(".", "", 1).isdigit()` para admitir decimales válidos y retorna `-1` ante entradas no numéricas. |
| `leer_opcion(mensaje)` | Valida que la selección ingresada sea un número entero mediante `.isdigit()`, devolviendo `-1` en caso de error. |
| `mostrar_titulo(banco)` | Formatea e imprime el encabezado del banco utilizando el método `banco.get_nombre().upper()`. |
| `cargar_datos_de_ejemplo(banco)` | Facilita pruebas y demostraciones poblando el sistema con clientes, cuentas de ahorros y cuentas corrientes con saldos iniciales. |



### 5.3. Funciones de Operaciones e Integración con POO
Cada opción del menú delega la lógica de negocio a los objetos instanciados, protegiendo la encapsulación de los datos:

* **Registro y Apertura:**
  * `registrar_cliente(banco)`: Captura DNI, nombres, apellidos y teléfono para invocar el registro en la instancia del banco.
  * `abrir_cuenta(banco)`: Solicita el DNI del titular, selecciona el subtipo de cuenta (1: Ahorros, 2: Corriente) y asigna el depósito inicial validado.
* **Transacciones Financieras:**
  * `depositar(banco)` y `retirar(banco)`: Buscan la cuenta por su identificador. Si existe, invocan los métodos polimórficos de la cuenta (`depositar()` o `retirar()`), informando el saldo actualizado mediante `cuenta.get_saldo()`.
  * `transferir(banco)`: Solicita cuenta de origen, cuenta de destino y monto, delegando la transacción directamente al método `banco.transferir()`.
* **Consultas y Cierres:**
  * `estado_de_cuenta(banco)`: Permite imprimir el historial detallado de movimientos de una cuenta específica a través de su método `estado_de_cuenta()`.
  * Operaciones bancarias globales: Opciones directas para `banco.mostrar_clientes()`, `banco.cierre_de_mes()` (cálculo de intereses y comisiones) y `banco.reporte_general()`.


### 5.4. Flujo de Control Principal (`main`)
La función `main()` implementa un bucle `while opcion != 0` que mantiene el sistema en ejecución continua hasta que el usuario decida salir (opción `0`). El menú evalúa la entrada a través de condicionales estructuradas (`if-elif-else`), manejando opciones fuera de rango con mensajes informativos sin interrumpir la ejecución del programa.

## 6. Conclusiones
La implementación de este sistema demuestra que el uso de la Programación Orientada a Objetos permite modelar escenarios del mundo real de forma intuitiva. El encapsulamiento fue vital para proteger el dinero de las cuentas, asegurando que las transacciones solo ocurran bajo las reglas permitidas.

Gracias a la aplicación de la Herencia, el código es altamente escalable. Si en el futuro el banco decide crear una "Cuenta Sueldo", bastará con crear una nueva subclase que herede de `Cuenta` sin necesidad de reescribir toda la lógica de retiros y depósitos.

La resolución de conflictos mediante Git y GitHub durante el desarrollo del proyecto evidenció la importancia de modularizar el código y separar las responsabilidades, permitiendo que distintos integrantes del grupo programaran el Banco, los Clientes y las Cuentas de forma paralela y las integraran exitosamente.
